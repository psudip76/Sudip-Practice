```python
import numpy as np
x=int(input("enter a number"))
def advanced_numpy_calculator(val):
 if(val<0):
   y=(np.exp(val)*np.cos(val))
   print("Damped Oscillation is : ",y)
 elif(0<=val<15):
   z=np.radians(val)
   y=(np.sin(2 * z)+np.cos(3 * z))
   print("Trigonometric Combination is: ",y)
 elif(15<=val<40):
   y=(np.log1p(val)*np.sqrt(val)-np.cbrt(val))
   print("Logarithmic Root Scaling: ",y)
 elif (40<=val<90):
   new_arr=np.array([[val,val/2,val/3],
                 [val/4,val/5,val/6],
                 [val/7,val/8,val/9]])
   y=np.linalg.det(new_arr)
   print("matrix determinant is: ",y)
 elif(90<=val<150):
   coeffs=[1,-3,2,-1,5]
   y=np.polyval(coeffs,val)
   print("Polynomial Evaluation: ",y)
 else:
   test_arr=np.array([[val,val/2,val/3,val/4,val/5]])
   mean_i=np.mean(test_arr)
   std_i=np.std(test_arr)
   y =mean_i+std_i
   print("Mean+standard deviation =", y)
advanced_numpy_calculator(x)
```
